```python
import os
from style import *
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
fcquic_path = "../out/npf_out.csv"
baseline_path = "../out/npf_out_baseline.csv"
```

```python
df_fcquic = pd.read_csv(fcquic_path)
display(df_fcquic)
len_fcquic = len(df_fcquic["y_LATENCY"])
print(f"length of fcquic results: {len_fcquic}")

df_baseline = pd.read_csv(baseline_path)
display(df_baseline)
len_baseline = len(df_baseline["y_LATENCY"])
print(f"length of baseline results: {len_baseline}")

global_len = min(len_fcquic, len_baseline)
df_fcquic = df_fcquic[:global_len]
df_basline= df_baseline[:global_len]

delay_str = str(df_fcquic["DELAY"][0])
fb_delay_str = str(df_fcquic["FALLBACK_DELAY"][0])
loss_str = str(df_fcquic["LOSS_RATE"][0]).replace("%", "\%")
burst_str = str(df_fcquic["BURST_RATE"][0]).replace("%", "\%")
```

```python
sns.set_style("whitegrid")
plt.figure(figsize=(8, 6))

# np.arange(1, len(sorted_values) + 1) / len(sorted_values)
x_values = np.arange(1, global_len + 1) / global_len
df = pd.DataFrame()
# df["x"] = x_values
df["fcquic"] = df_fcquic["y_LATENCY"] / 1000
df["baseline"] = df_baseline["y_LATENCY"] / 1000

# sns.ecdfplot(df, palette='Set2', legend=True)
latexify(nb_subplots_line=1, fig_height=8, fig_width=6)
plt.ecdf(df["fcquic"], label="FC-QUIC", color=COLORS[1])
plt.ecdf(df["baseline"], label="Baseline QUIC", color=COLORS[2])
plt.ylabel("Probability", fontsize=12)
plt.title("Cumulative Distribution Function of latency", fontsize=14)
plt.xlabel("Latency (ms)", fontsize=12)
plt.figtext(0.51, -0.01, f"Delay: {delay_str}, loss rate: {loss_str}, burst rate: {burst_str}, fallback delay: {fb_delay_str} ms", ha="center", fontsize=12)
# plt.xlim(left=0)
plt.ylim(0, 1)
plt.legend()
plt.grid(True, alpha=0.3)


plt.tight_layout()
plt.savefig("cdf_both.png", dpi=350, bbox_inches="tight")
plt.savefig("cdf_both.svg", bbox_inches="tight")

plt.show()
```
